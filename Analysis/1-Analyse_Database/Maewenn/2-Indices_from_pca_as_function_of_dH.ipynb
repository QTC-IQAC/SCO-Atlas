```python
sorted_indices_200D =[]
for i in range (len(systems_homoleptics)):
    idx = systems_homoleptics[i].ligands[0].sorted_index_200D_and_dist
    sorted_indices_200D.append(idx)
```

```python
# Initialize a list to store sorted ligands and their corresponding dH values
sorted_ligands_info = []

# Loop through systems_homoleptics to extract sorted ligands and their dH values
count = 0
for i in sorted_indices_200D:
    # Access the ligand and its relevant properties (adjust the attributes according to your data structure)
    ligand = systems_homoleptics[i].ligands[0]
    smiles = ligand.smiles  # Adjust to the actual SMILES or other relevant property
    # Replace the SMILES string with the refcode
    refcode = systems_homoleptics[i].ligands[0].get_parent("cell").refcode
    # Round the dH value to 2 decimal places and add the ligand info to the list
    rounded_dH = round(dH[i], 2)
    # Stores position in ligand
    ligand.index_mols2grid = count
    count += 1
    # Add the ligand and its corresponding dH value to the list
    sorted_ligands_info.append({
        'SMILES': smiles,
        'Refcode': refcode,
        'dH': rounded_dH
    })

# Convert the list of sorted ligands info into a pandas DataFrame
sorted_ligands_df = pd.DataFrame(sorted_ligands_info)

# Now, display the DataFrame (this will show Name, SMILES, and dH columns)
import mols2grid
mols2grid.display(
    sorted_ligands_df,
    tooltip=['Refcode', 'dH'],
    subset=["Refcode", "dH"],
    style={"__all__": lambda x: "background-color: azure;"}  # Optional styling
)
```

```python
# Initialize a list to store ligands and their corresponding dH values
ligands_info = []

# Loop through systems_homoleptics to extract ligands and their dH values
for i in range(len(systems_homoleptics)):
    # Access the ligand and its relevant properties (adjust the attributes according to your data structure)
    ligand = systems_homoleptics[i].ligands[0]
    smiles = ligand.smiles  # Adjust to the actual SMILES or other relevant property
    # Replace the SMILES string with the refcode
    refcode = systems_homoleptics[i].ligands[0].get_parent("cell").refcode
    # Round the dH value to 2 decimal places and add the ligand info to the list
    rounded_dH = round(dH[i], 2)
    # Add the ligand and its corresponding dH value to the list
    ligands_info.append({
        'SMILES': smiles,
        'Refcode': refcode,
        'dH': rounded_dH
    })

# Sort the ligands based on dH
sorted_ligands_info = sorted(ligands_info, key=lambda x: x['dH'])

# Convert the list of sorted ligands info into a pandas DataFrame
sorted_ligands_df = pd.DataFrame(sorted_ligands_info)

# Now, display the DataFrame (this will show Name, SMILES, and dH columns)
import mols2grid
mols2grid.display(
    sorted_ligands_df,
    tooltip=['Refcode', 'dH'],
    subset=["Refcode", "dH"],
    style={"__all__": lambda x: "background-color: azure;"}  # Optional styling
)
```
