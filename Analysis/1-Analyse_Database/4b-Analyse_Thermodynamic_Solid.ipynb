```python
import scope
import scope_sco
import os
import csv
import matplotlib.pyplot as plt
import numpy as np
```

```python
def get_values(csv_path, column):
    values = []
    with open(csv_path, newline="") as f:
        reader = csv.DictReader(f)
        for row in reader:
            values.append(row[column])
    return values
```

```python
finished_csv_path     = outputs_path + "finished_systems.csv"

finished_systems = []
not_finished = []
with open(finished_csv_path, newline="") as f:
    reader = csv.DictReader(f)
    for row in reader:
        sys = scope.read_write.load_binary(f"{row['path']}")
        sys.set_paths_from_environment(env)
        finished_systems.append(sys)
```

```python
systems_with_solid_state = []
for sys in finished_systems:
    found, solid_branch = sys.find_branch("solid")
    if found and sys.refcode not in new_ss_to_be_computed:
        last_job = sys.branches[-1].workflows[-1].jobs[-1]
        if last_job.name == "vc" and last_job.isgood:
            print(sys.refcode, "not in ss list but finished and good")
    #elif found and sys.refcode in new_ss_to_be_computed:
    #    print(sys.refcode + " is in the list of systems to be computed")
```

```python
for sys in systems_with_solid_state:
    try:
        found1, HS_state1 = find_state(sys.HS_ref_mol,"initial")
        found2, LS_state1 = find_state(sys.LS_ref_mol,"initial")
        found3, HS_state2 = find_state(sys.HS_ref_mol,"b3lyp_opt")
        found4, LS_state2 = find_state(sys.LS_ref_mol,"b3lyp_opt")
        found5, HS_state3 = find_state(sys.HS_ref_crys.cell,"relax")
        found6, LS_state3 = find_state(sys.LS_ref_crys.cell,"relax")

        if not found1 or not found2 or not found3 or not found4 or not found5 or not found6:
            print(sys.refcode, "not relax state")
            continue
            
        HS_mol1 = HS_state1.get_moleclist()[0]
        HS_mol2 = HS_state2.get_moleclist()[0]
        LS_mol1 = LS_state1.get_moleclist()[0]
        LS_mol2 = LS_state2.get_moleclist()[0]

        found = False
        for mol in HS_state3.get_moleclist():
            if mol == HS_mol1 and not found:
                #writexyz('/Users/sergivela/Documents/SCOPE/Program/tests/Overlap_Molecules/', "mol.xyz", mol.labels, mol.coord)
                #writexyz('/Users/sergivela/Documents/SCOPE/Program/tests/Overlap_Molecules/', "hsmol1.xyz", HS_mol1.labels, HS_mol1.coord)
                val = mol.rmsd(HS_mol1, reorder=True, center_method="metal", debug=0)
                if val is not None: 
                    if val < 0.2: 
                        found = True
                        HS_mol3 = mol

        found = False
        for mol in LS_state3.get_moleclist():
            if mol == LS_mol1 and not found:
                val = mol.rmsd(LS_mol1, reorder=True, center_method="metal", debug=0)
                if val is not None: 
                    if val < 0.2: 
                        found = True
                        LS_mol3 = mol

        print(sys.refcode, HS_mol1.rmsd(HS_mol2), HS_mol1.rmsd(HS_mol3), LS_mol1.rmsd(LS_mol2), LS_mol1.rmsd(LS_mol3))
    except Exception as e:
        print(sys.refcode, e)
```
