```python
for sys in finished_systems:
    if sys.refcode == "NAPYFE10":
        for br in sys.branches:
            if   br.keyword == 'Isolated': exts = ['.com', '.log', '.sub'] 
            elif br.keyword == 'Solid':    exts = ['.input', '.out', '.sub'] 
            for rec in br.recipes:
                for job in rec.jobs:
                    for comp in job.computations:
                        print("processing", comp.inp_path, comp.isregistered)
                        #if not comp.isregistered: print("not_registered", comp.inp_path); continue
                        if comp.step > 1: has_step = True
                        else:             has_step = False
    
                        if   br.keyword == 'Solid'    and rec.subject.spin == 'HS': workflow = 'ref_hs_cell'
                        elif br.keyword == 'Solid'    and rec.subject.spin == 'LS': workflow = 'ref_ls_cell'
                        elif br.keyword == 'Isolated' and rec.subject.spin == 'HS': workflow = 'ref_hs_mol'
                        elif br.keyword == 'Isolated' and rec.subject.spin == 'LS': workflow = 'ref_ls_mol'

                        if has_step: new_file = f"{sys.refcode}_{workflow}_{job.suffix}_s{comp.step}_r{comp.run_number}"
                        else:        new_file = f"{sys.refcode}_{workflow}_{job.suffix}_r{comp.run_number}"

                        new_path = f"{new_comps_folder}/{sys.refcode}/{br.keyword}/{new_file}"

                        alter = False
                        os.makedirs(f"{new_comps_folder}/{sys.refcode}/{br.keyword}", exist_ok=True)

                        #Input
                        if not os.path.exists(comp.inp_path): 
                            print("missing", comp.inp_path) 
                        elif not os.path.exists(f"{new_path}{exts[0]}"):
                            print("copying", comp.inp_path, f"{new_path}{exts[0]}")
                            shutil.copy2(comp.inp_path, f"{new_path}{exts[0]}")

                        #Output    
                        if not os.path.exists(comp.out_path) and not os.path.exists(f"{comp.out_path}_part"): print("missing", comp.out_path) 
                        else: 
                            if os.path.exists(comp.out_path):  
                                print("copying", comp.out_path, f"{new_path}{exts[1]}")
                                shutil.copy2(comp.out_path, f"{new_path}{exts[1]}") 
                            else:
                                print("copying", f"{comp.out_path}_part", f"{new_path}{exts[1]}")
                                shutil.copy2(f"{comp.out_path}_part", f"{new_path}{exts[1]}") 

                        #Submission
                        if not os.path.exists(comp.sub_path): 
                            print("missing", comp.sub_path) 
                        else: 
                            print("copying", comp.sub_path, f"{new_path}{exts[2]}")
                            shutil.copy2(comp.sub_path, f"{new_path}{exts[2]}")
```
